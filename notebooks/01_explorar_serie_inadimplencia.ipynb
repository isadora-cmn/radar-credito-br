```python
import requests
import pandas as pd

url_serie_inadimplencia = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.21082/dados?formato=json"

resposta_banco_central = requests.get(url_serie_inadimplencia, timeout=30)
dados_em_json = resposta_banco_central.json()

tabela_inadimplencia = pd.DataFrame(dados_em_json)
tabela_inadimplencia.head()
```

```python
tabela_inadimplencia["data"] = pd.to_datetime(tabela_inadimplencia["data"], format="%d/%m/%Y")
tabela_inadimplencia["valor"] = pd.to_numeric(tabela_inadimplencia["valor"])

tabela_inadimplencia.info()
```

```python
tabela_inadimplencia.tail()
```

```python
import plotly.express as px

grafico_inadimplencia_total = px.line(
    tabela_inadimplencia,
    x="data",
    y="valor",
    title="Inadimplência total do crédito no Brasil (% da carteira)",
    labels={"data": "Mês", "valor": "Inadimplência (%)"},
)
grafico_inadimplencia_total.show()
```
